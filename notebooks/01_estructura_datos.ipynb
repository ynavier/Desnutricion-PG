### Cargar los datasets desde data/external y data/raw


```python
bucket_url = "gs://data_desnutricion"
file_names = {
    "2020": "2020_113_Reporte.xlsx",
    "2021": "2021_113_Reporte.xlsx",
    "2022": "2022_113_Reporte.xlsx",
    "2023": "2023_113_Reporte.xlsx",
    "2024": "2024_113_Reporte.xlsx",
    "2025": "2025_113_Reporte.xlsx",
}

datasets = {}
for year, file_name in file_names.items():
    url = f"{bucket_url}/{file_name}"
    print(f"Cargando {year} desde {url}")
    datasets[year] = pd.read_excel(
        url,
        engine="openpyxl",
        storage_options={"token": "google_default"}
    )
    print(f"{year}: {datasets[year].shape[0]:,} filas x {datasets[year].shape[1]} columnas")
```

### Revisar dimensiones de los datasets

```python
for year, df in datasets.items():
    print(f"\n===== DIMENSIONES DATASET {year} =====")
    print("Filas y columnas:", df.shape)
```

### Revisar tipos de datos

```python
for year, df in datasets.items():
    print(f"\n===== TIPOS DE DATOS {year} =====")
    print(df.dtypes)
```

### Ver columnas de cada dataset

```python
for year, df in datasets.items():
    print(f"\n===== COLUMNAS DATASET {year} =====")
    print(list(df.columns))
```

### Revisar valores nulos

```python
for year, df in datasets.items():
    print(f"\n===== NULOS DATASET {year} =====")
    print(df.isnull().sum().sort_values(ascending=False))
```

### Ver solo columnas con nulos

```python
for year, df in datasets.items():
    print(f"\n===== COLUMNAS CON NULOS {year} =====")
    
    nulls = df.isnull().sum()
    nulls = nulls[nulls > 0].sort_values(ascending=False)
    
    print(nulls)
```

### Porcentaje de datos faltantes

```python
for year, df in datasets.items():
    print(f"\n===== PORCENTAJE DE NULOS {year} =====")
    
    percent = (df.isnull().sum() / len(df)) * 100
    print(percent.sort_values(ascending=False))
```

### Revisar duplicados

```python
for year, df in datasets.items():
    print(f"\n===== DUPLICADOS DATASET {year} =====")
    print("Registros duplicados:", df.duplicated().sum())
```

### Ver primeras filas

```python
for year, df in datasets.items():
    print(f"\n===== PRIMERAS FILAS DATASET {year} =====")
    display(df.head())
```

### Comparar si las columnas son iguales entre datasets

```python
common_cols = set.intersection(*(set(df.columns) for df in datasets.values()))
print("Columnas comunes a todos los años:", len(common_cols))
print(sorted(common_cols)[:30])
for year, df in datasets.items():
    unique_cols = set(df.columns) - common_cols
    print(f"{year}: columnas únicas = {len(unique_cols)}")
    if unique_cols:
        print(sorted(unique_cols)[:10])
```

# ====== Variable Objetivo ======

```python
for year, df in datasets.items():
    print(f"\n=== {year} ===")
    for col in df.columns:
        print(col, "-", df[col].nunique())
```

### filtramos columnas con pocos valores únicos, que suelen ser targets en clasificación.

```python
for year, df in datasets.items():
    print(f"\n=== {year} ===")
    for col in df.columns:
        unique_vals = df[col].nunique()
        if 2 <= unique_vals <= 10:
            print(col, "-", unique_vals)
```
